```python
import pandas as pd
import numpy as np

from pathlib import Path
from tqdm import tqdm
```

```python
PROJECT_DIR = Path("..")

RAW_DIR = PROJECT_DIR / "data" / "raw"
PROCESSED_DIR = PROJECT_DIR / "data" / "processed"

META_DIR = RAW_DIR / "meta_data"
SMART_DIR = RAW_DIR / "smart_meter_data"
WEATHER_DIR = RAW_DIR / "weather_data"
```

```python
print("PROJECT :", PROJECT_DIR.resolve())
print("RAW :", RAW_DIR.resolve())

print("META :", META_DIR.exists())
print("SMART :", SMART_DIR.exists())
print("WEATHER :", WEATHER_DIR.exists())
print("PROCESSED :", PROCESSED_DIR.exists())
```

Load Household Terpilih

```python
selected_households = pd.read_csv(
    PROCESSED_DIR /
    "selected_households.csv"
)

selected_households.head()
```

```python
selected_ids = (
    selected_households[
        "Household_ID"
    ].tolist()
)

print(
    "Jumlah household:",
    len(selected_ids)
)
```

Load Metadata

```python
meta_df = pd.read_csv(
    META_DIR /
    "meta_data.csv",
    sep=";"
)

households_df = pd.read_csv(
    META_DIR /
    "households.csv",
    sep=";"
)
```

```python
print("Metadata :", meta_df.shape)
print("Households :", households_df.shape)
```

Filter Hanya 300 Household

```python
meta_df = meta_df[
    meta_df["Household_ID"]
    .isin(selected_ids)
]

households_df = households_df[
    households_df["Household_ID"]
    .isin(selected_ids)
]
```

```python
print(
    "Metadata selected:",
    meta_df.shape
)

print(
    "Households selected:",
    households_df.shape
)
```

Load Smart Meter 300 Household

```python
SMART_15MIN_DIR = (
    SMART_DIR /
    "15min"
)

smart_files = sorted(
    SMART_15MIN_DIR.glob("*.csv")
)

len(smart_files)
```

```python
for f in smart_files:
    print(f.name)
```

```python
smart_list = []

for file in tqdm(smart_files):

    temp = pd.read_csv(
        file,
        sep=';'
    )

    temp = temp[
        temp['Household_ID']
        .isin(selected_ids)
    ]

    smart_list.append(temp)
```

```python
smart_df = pd.concat(
    smart_list,
    ignore_index=True
)
```

```python
smart_df.head()
```

```python
smart_df.info()
```

```python
smart_df['Timestamp'] = pd.to_datetime(
    smart_df['Timestamp'],
    utc=True
)
```

```python
smart_df = (
    smart_df
    .sort_values(
        ['Household_ID', 'Timestamp']
    )
    .reset_index(drop=True)
)
```

```python
len(smart_df)
```

```python
weather_files = sorted(
    (WEATHER_DIR / "hourly").glob("*.csv")
)

len(weather_files)
```

Kesimpulan Tahap 01 – Data Integration

Pada tahap Data Integration, seluruh sumber data yang relevan berhasil digabungkan menjadi satu dataset utama (master_df) yang siap digunakan untuk proses Exploratory Data Analysis (EDA) dan preprocessing. Integrasi dilakukan secara bertahap dengan menggabungkan data smart meter, metadata rumah tangga, dan data cuaca menggunakan kunci yang sesuai, yaitu Household_ID untuk menghubungkan data rumah tangga dan Weather_ID serta Timestamp_hour untuk menghubungkan data cuaca.

Proses integrasi dimulai dengan pemilihan 300 household terbaik berdasarkan kualitas data, yang mempertimbangkan durasi observasi, tingkat missing value, dan proporsi konsumsi nol. Selanjutnya, data metadata (meta_data.csv) dan data karakteristik rumah tangga (households.csv) difilter agar hanya mencakup 300 household terpilih. Setelah itu, seluruh data smart meter 15-menit dari 300 household digabungkan sehingga menghasilkan 30.735.648 observasi yang merepresentasikan konsumsi energi rumah tangga selama periode pengamatan.

Tahap berikutnya adalah mengintegrasikan informasi karakteristik rumah tangga ke dalam data smart meter melalui Household_ID. Hasil penggabungan menunjukkan bahwa seluruh observasi berhasil memperoleh informasi metadata tanpa menghasilkan duplikasi data. Setelah itu, data cuaca per jam diintegrasikan menggunakan kombinasi Weather_ID dan Timestamp_hour, sehingga setiap observasi konsumsi energi memiliki informasi kondisi lingkungan pada waktu yang sama.

Hasil evaluasi integrasi menunjukkan bahwa variabel cuaca hanya memiliki tingkat missing value yang sangat kecil (kurang dari 0,5% dari total observasi), sehingga tidak menimbulkan masalah serius terhadap kualitas data. Selain itu, proses merge tidak menghasilkan penambahan jumlah baris maupun duplikasi Household_ID, yang menunjukkan bahwa integrasi data telah dilakukan secara konsisten dan valid.

Secara keseluruhan, tahap Data Integration berhasil menghasilkan sebuah master dataset yang terintegrasi, konsisten, dan siap untuk dianalisis lebih lanjut. Dataset ini telah menggabungkan informasi konsumsi energi, karakteristik rumah tangga, dan kondisi cuaca dalam satu struktur data terpadu, sehingga memungkinkan dilakukannya analisis eksploratif yang lebih mendalam pada tahap berikutnya, yaitu 02 – Exploratory Data Analysis (EDA) untuk menentukan strategi preprocessing dan feature engineering yang paling tepat.
